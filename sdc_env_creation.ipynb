### System dependencies for LaTeX rendering
SciencePlots enables `text.usetex = True`, which routes text rendering through the system LaTeX pipeline. Two system packages are required:
- `cm-super`: provides `type1ec.sty` (font package) needed by LaTeX
- `dvipng`: converts `.dvi` output to `.png` so matplotlib can embed it

These are **system** (apt) packages, not conda/pip packages.

```python
!sudo apt-get install -y cm-super dvipng
```
