In our case we torch==1.12.0 doesn't support sm_86 (RTX 3090). To fix it for CUDA 12.2 we need to stay compatible with PyTorch 1.12 which supports up to CUDA 11.6.                            

```python
%%bash 
conda create --name shifts_mswml python=3.8.10 -y
source activate shifts_mswml 
pip install -r shifts/mswml/requirements.txt
pip install torch==1.12.0+cu116 torchvision==0.13.0+cu116 \
    --extra-index-url https://download.pytorch.org/whl/cu116
export VAR="shifts_mswml"
```
