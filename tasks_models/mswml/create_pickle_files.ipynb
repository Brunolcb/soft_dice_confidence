```python
import numpy as np
import nibabel as nib
dev_in_num = 7
dev_out_num = 25
eval_in_num = 33
train_num = 33

dev_in_path_gt = ['./shifts2_data/dev_in/gt/%d_gt_isovox.nii.gz'%(i+1) for i in range(dev_in_num)]
dev_out_path_gt = ['./shifts2_data/dev_out/gt/%d_gt_isovox.nii.gz'%(i+1) for i in range(dev_out_num)]
eval_in_path_gt = ['./shifts2_data/eval_in/gt/%d_gt_isovox.nii.gz'%(i+1) for i in range(eval_in_num)]
train_path_gt = ['./shifts2_data/train/gt/%d_gt_isovox.nii.gz'%(i+1)for i in range(train_num)]

dev_in_path_predictions = ['./predictions/predictions_dev_in/%d_pred_prob.nii.gz'%(i+1) for i in range(dev_in_num)]
dev_out_path_predictions = ['./predictions/predictions_dev_out/%d_pred_prob.nii.gz'%(i+1) for i in range(dev_out_num)]
eval_in_path_predictions = ['./predictions/predictions_eval_in/%d_pred_prob.nii.gz'%(i+1) for i in range(eval_in_num)]
train_path_predictions = ['./predictions/predictions_train/%d_pred_prob.nii.gz'%(i+1)for i in range(train_num)]
```

```python
dev_in_path_gt
```

```python
def nifiti_to_array(path):
    img = nib.load(path)
    a = np.array(img.dataobj)
    return np.expand_dims(a, axis=0)
```

```python
a_dev_in_path_gt = [nifiti_to_array(i) for i in dev_in_path_gt]
a_dev_out_path_gt = [nifiti_to_array(i) for i in dev_out_path_gt]
a_eval_in_path_gt = [nifiti_to_array(i) for i in eval_in_path_gt]
a_train_path_gt = [nifiti_to_array(i) for i in train_path_gt]

a_dev_in_path_predictions = [nifiti_to_array(i) for i in dev_in_path_predictions]
a_dev_out_path_predictions = [nifiti_to_array(i) for i in dev_out_path_predictions]
a_eval_in_path_predictions = [nifiti_to_array(i) for i in eval_in_path_predictions]
a_train_path_predictions = [nifiti_to_array(i) for i in train_path_predictions]
```

```python
import pickle
```

```python
with open('dev_in_gt.pkl', 'wb') as f:
    pickle.dump(a_dev_in_path_gt, f)
    
with open('dev_out_gt.pkl', 'wb') as f:
    pickle.dump(a_dev_out_path_gt, f)
    
with open('eval_in_gt.pkl', 'wb') as f:
    pickle.dump(a_eval_in_path_gt, f)

with open('train_gt.pkl', 'wb') as f:
    pickle.dump(a_train_path_gt, f)
```

```python
with open('dev_in_predictions.pkl', 'wb') as f:
    pickle.dump(a_dev_in_path_predictions, f)
    
with open('dev_out_predictions.pkl', 'wb') as f:
    pickle.dump(a_dev_out_path_predictions, f)
    
with open('eval_in_predictions.pkl', 'wb') as f:
    pickle.dump(a_eval_in_path_predictions, f)

with open('train_predictions.pkl', 'wb') as f:
    pickle.dump(a_train_path_predictions, f)
```

```python
import os
import shutil

dest_dir = os.path.join('..', '..',"medical-imaging", 'data', 'mswml')
os.makedirs(dest_dir, exist_ok=True)

pickle_files = [
    'dev_in_gt.pkl',
    'dev_out_gt.pkl',
    'eval_in_gt.pkl',
    'train_gt.pkl',
    'dev_in_predictions.pkl',
    'dev_out_predictions.pkl',
    'eval_in_predictions.pkl',
    'train_predictions.pkl',
]

for fname in pickle_files:
    if os.path.exists(fname):
        shutil.move(fname, os.path.join(dest_dir, fname))
        print(f'Moved {fname} -> {dest_dir}/')
    else:
        print(f'Skipped (not found): {fname}')
```
