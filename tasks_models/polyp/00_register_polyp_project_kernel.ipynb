# Create and register Jupyter kernel — polyp_project

Run this notebook **once** with any available kernel (e.g.: `base` / `Python 3`).
The steps below will:
1. Create the conda `polyp_project` environment (Python 3.10)
2. Install PyTorch 2.x with CUDA 12.1 support (compatible with CUDA 12.2)
3. Install `timm==0.6.13`, `torchvision`, `numpy`, `Pillow`, `tqdm`, `ipykernel`
4. Register the `polyp_project` kernel in Jupyter

After running, restart Jupyter and the `polyp_project` kernel will be available for selection.

```python
import subprocess, sys

ENV_NAME = 'polyp_project'

def run(cmd, desc):
    print(f'\n=== {desc} ===')
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.stdout.strip():
        print(result.stdout[-3000:])
    if result.returncode != 0:
        print('STDERR:', result.stderr[-2000:])
        raise RuntimeError(f'Failed: {desc}')
    print(f'OK — {desc}')

# ── Step 1: Create conda environment ─────────────────────────────────────────────
run(
    ['conda', 'create', '-n', ENV_NAME, 'python=3.10', '-y'],
    f"Criar ambiente conda '{ENV_NAME}'"
)
```

```python
# ── Step 2: PyTorch + torchvision with CUDA 12.1 ──────────────────────────────
# CUDA 12.1 wheels are compatible with the CUDA 12.2 driver installed on the machine.
run(
    ['conda', 'run', '-n', ENV_NAME, 'pip', 'install',
     'torch', 'torchvision',
     '--index-url', 'https://download.pytorch.org/whl/cu121'],
    'Install PyTorch + torchvision (CUDA 12.1)'
)
```

```python
# ── Step 3: Other dependencies ──────────────────────────────────────────────
# timm==0.6.13: version compatible with pvtv2.py imports (registry, _cfg, layers)
run(
    ['conda', 'run', '-n', ENV_NAME, 'pip', 'install',
     'timm==0.6.13', 'numpy', 'Pillow', 'tqdm', 'ipykernel'],
    'Install timm, numpy, Pillow, tqdm, ipykernel'
)
```

```python
# ── Step 4: Register Jupyter kernel ─────────────────────────────────────────
run(
    ['conda', 'run', '-n', ENV_NAME,
     'python', '-m', 'ipykernel', 'install',
     '--user', '--name', ENV_NAME, '--display-name', ENV_NAME],
    f"Registrar kernel '{ENV_NAME}' no Jupyter"
)

print('\nAll done!')
print('Restart Jupyter and select "polyp_project" from the kernel list.')
```
