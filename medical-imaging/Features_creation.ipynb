## AEF's Features creations
#### Note: This notebook should be run only after `Optimum_uncertainty_thresholds.ipynb`.

```python
from src.metrics import dice_coef
from src.utils import entropy_func, uncertainty_error_overlap, feature_extractor_3D
from collections import defaultdict
import numpy as np
import pickle
```

In this notebook we will create the features from the optimium uncertainty thresholds in the tuning sets that will be used to train the Random Forest for AEF confidence estimator. Additionally, the Dice value for each prediction, label pair will be computed to be used later as the target for the same Random Forest model.

```python
task ='mswml_ood'
if 'mswml' not in task:
    data = np.load(f'./data/{task}.npz')
    y = data['y']
    p_hat = data['p_hat']
    y_hat = data['y_hat'] # This line should be COMMENTED in case y_hat is not given
elif 'ood' in task:
    p_hat = pickle.load(open('./data/mswml/dev_out_predictions.pkl', 'rb'))
    y = pickle.load(open('./data/mswml/dev_out_gt.pkl', 'rb'))
else:
    p_hat = pickle.load(open('./data/mswml/eval_in_predictions.pkl', 'rb'))
    y = pickle.load(open('./data/mswml/eval_in_gt.pkl', 'rb'))
```

```python
uncertainty_masks = entropy_func(p_hat)
```

```python
with open(f'./data/{task}_u_threshs.pkl', 'rb') as file:
    # SLoad the unique list to the file
    u_threshs =  np.unique(pickle.load(file))
```

```python
features_dict = defaultdict(list)
for u_thresh in u_threshs:
    for msk in uncertainty_masks:
        features = feature_extractor_3D(msk,u_thresh=u_thresh)
        features_dict[u_thresh].append(features)
```

```python
with open(f'./data/{task}_features_dict.pkl', 'wb') as file:
    # Save the list to the file
    pickle.dump(features_dict, file)
```

```python
targets = []

# If y_hat is not given, use this code
for y_i,p_hat_i in zip(y, p_hat):
    targets.append(dice_coef(p_hat_i,y_i, threshold=0.35))

#When y_hat is given (BraTS)
#for y_i,y_hat_i in zip(y, y_hat):
#    targets.append(dice_coef(y_hat_i,y_i))
```

```python
import pickle
with open(f'./data/{task}_targets.pkl', 'wb') as file:
    # Save the list to the file
    pickle.dump(targets, file)
```
